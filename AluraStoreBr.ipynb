### Importação dos dados



```python
import pandas as pd
!pip install -q numpy==1.23.5

url1 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja1 = pd.read_csv(url1)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja1.head()
```

#1. Análise do faturamento


```python
# Loja 1
faturamento_loja1 = loja1['Preço'].sum()

# Loja 2
faturamento_loja2 = loja2['Preço'].sum()

# Loja 3
faturamento_loja3 = loja3['Preço'].sum()

# Loja 4
faturamento_loja4 = loja4['Preço'].sum()

# Criando lista de dados para montar o gráfico
lista_lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']
lista_faturamentos = [faturamento_loja1, faturamento_loja2, faturamento_loja3, faturamento_loja4]

# Importando a biblioteca e criando o gráfico
!pip install -q matplotlib
import matplotlib.pyplot as plt

# Criando o gráfico
plt.bar(lista_lojas, lista_faturamentos, color='mediumseagreen')

# Títulos e rotulos
plt.title('Faturamento das Lojas')
plt.xlabel('Lojas')
plt.ylabel('Faturamento (R$)')

# Mostrar valores em cima das barras
for i, valor in enumerate(lista_faturamentos):
  plt.text(i, valor + 10, f'R$ {valor:,.2f}'.replace(',', 'v').replace('.', ',').replace('v', '.'), ha='center')

# Mostrar o gráfico
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

# 2. Vendas por Categoria


```python
# Faturamento por categoria Loja 1
faturamento_por_categoria = loja1.groupby('Categoria do Produto')['Preço'].sum().sort_values(ascending=False)
faturamento_por_categoria

#Criar o gráfico
explode = [0.1 if valor < 20000 else 0 for valor in faturamento_por_categoria]
fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(
    faturamento_por_categoria,
    labels=None,
    autopct='%1.0f%%',
    startangle=90,
    explode=explode,
    colors=plt.cm.Paired.colors,
    textprops={'fontsize': 7},
)

ax.set_position([0.05, 0.05, 0.75, 0.9])

plt.legend(
    [f'{cat} - R$ {valor:,.0f}'.replace(',', '.')
     for cat, valor in zip(faturamento_por_categoria.index, faturamento_por_categoria.values)],
    title='Categorias',
    bbox_to_anchor=(1.05, 1),
    loc='upper left'
)

plt.title('Faturamento por Categoria da Loja 1')
plt.show()
```

# 3. Média de Avaliação das Lojas

```python
# Média de avaliações Loja 1
media_avaliacoes_loja1 = loja1['Avaliação da compra'].mean()
print(f'Média das avaliações da Loja 1: {media_avaliacoes_loja1:.2f}')

# Média de avaliações Loja 2
media_avaliacoes_loja2 = loja2['Avaliação da compra'].mean()
print(f'Média das avaliações da Loja 2: {media_avaliacoes_loja2:.2f}')

# Média de avaliações Loja 3
media_avaliacoes_loja3 = loja3['Avaliação da compra'].mean()
print(f'Média das avaliações da Loja 3: {media_avaliacoes_loja3:.2f}')

# Média de avaliações Loja 4
media_avaliacoes_loja4 = loja4['Avaliação da compra'].mean()
print(f'Média das avaliações da Loja 4: {media_avaliacoes_loja4:.2f}')

plt.bar(['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4'], [media_avaliacoes_loja1, media_avaliacoes_loja2, media_avaliacoes_loja3, media_avaliacoes_loja4])
plt.title('Média de Avaliação das Lojas')
plt.xlabel('Lojas')
plt.ylabel('Média de Avaliação')
plt.show()
```
